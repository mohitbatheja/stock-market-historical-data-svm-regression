```python
import pandas as pd
```

```python
df = pd.read_csv("Stock Market Historical Data.csv")
```

```python
df
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
df["Close/Last"] = (
    df["Close/Last"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```

```python
df["Close/Last"]
```

```python
df["Open"] = (
    df["Open"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```

```python
df["High"] = (
    df["High"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```

```python
df["Low"] = (
    df["Low"]
    .str.replace("$", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)
```

```python
df.info()
```

```python
df["Date"] = pd.to_datetime(df["Date"],errors = 'coerce')
```

```python
df["Date"].value_counts(dropna=False)
```

```python
df.isna().sum()
```

```python
df = df.dropna(subset=["Date"])
```

```python
df
```

```python
df.isna().sum()
```

```python
x = df.drop(["Company","Date", "Close/Last"], axis=1)
y = df["Close/Last"]
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error,mean_absolute_error,r2_score
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size = 0.20, random_state = 24
)
```

```python
scaler = StandardScaler()

x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)
```

```python
svm_r = SVR(
    kernel = "rbf"
)
```

```python
svm_r.fit(x_train,y_train)
```

```python
y_pred = svm_r.predict(x_test)
```

```python
y_pred
```

```python
compariosn = pd.DataFrame({
    "Actual" : y_test,
    "Prediction" : y_pred
})
```

```python
compariosn
```

```python
compariosn['Error'] = compariosn['Actual'] - compariosn['Prediction']
compariosn
```

```python
mae = mean_absolute_error(y_test,y_pred)
print("MAE is :",mae)
mse = mean_squared_error(y_test,y_pred)
print("MSE is :",mse)
rmse = np.sqrt(mse)
print("RMSE is :",rmse)
r2 = r2_score(y_test,y_pred)
print("R2 Score is :",r2)
```
